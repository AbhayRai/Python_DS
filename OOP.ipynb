#### Encapsulation

```python
class BankAccount:
    def __init__(self,name,Address):
        self.name = name;  ## Public variable
        self.Address = Address;
        pass

    def getName(self):
        return self.name
        
```

```python
BA = BankAccount('Abhay','Bangalore')
```

```python
BA.getName()
```

## Type of Varibale

```python
class TypeOfVariable:
    def  __init__(self,name,balance):
        self.name = name;
        self._balance = balance;
        self.__AccountType = 'Savings'
        pass
```

```python
BA1 = TypeOfVariable('Abhay',10000)
```

```python
BA1._balance    ## Protected
```

```python
BA1.__AccountType
```

```python
BA1.__AccountType = 'Recurring'
```

```python
BA1.__AccountType
```

```python
BA1._TypeOfVariable__AccountType      ## name mangling (For private variable can access outside of the class)
```

```python
BA1._TypeOfVariable__AccountType = 'New Account'
```

```python
## multiLevel Inheritance

class Device:
    def a(self):
        print('abc')
class TV(Device):
    def a2(self):
        print('abc')
class SmartBulb(TV):
    def a3(self):
        print('abc')
```

```python
SB = SmartBulb()
```

```python
SB.a()
SB.a2()
SB.a3()
```
