The map() Function in python

The map() function applies a given function to all items in input list (or any other iterable) and returns a map object (an iterator). This is particularly useful for transforming data in a list comprehensively.

```python
def square(x):
    return x*x
square(4)
```

```python
numbers=[1,2,3,4,5,5,6,7]

tuple(map(square,numbers))
```

```python
## lambda function with map
numbers=[1,2,3,4,5,6]
list(map(lambda x:x*x,numbers))
```

```python
### Map multiple iterables

numbers1=[1,2,3,4]
numbers2=[5,6,7]

added_numbers=list(map(lambda x,y:x+y,numbers1,numbers2))
print(added_numbers)
```
