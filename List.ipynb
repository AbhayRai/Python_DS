```python
## Create a list of cubes from 1 to 10

cube = []
for i in range(1,11):
    cube.append(i**3)
print(cube)
```

```python
## From a list [12, 7, 19, 4, 23], create a new list of only numbers > 10

data = [12, 7, 19, 4, 23]
Greater_Num = [i for i in data if i>10]
print(Greater_Num)
```

```python
## Convert a list of strings to uppercase using list comprehensions

List_of_String = ['abhay', 'rai', 'Python']

UpperCase_String = [i.upper() for i in List_of_String ]

print(UpperCase_String)
```

```python
## Use map to convert [1, 2, 3] → [1, 4, 9]
num = [1, 2, 3]

sqaured_num = (list(map(lambda x:x**2,num)))

print(sqaured_num)
```

```python
## Use filter to extract words > 3 letters

words = ["AI", "Data", "ML", "Analytics"]

Greater_Than_Three_Letters = list(filter(lambda x: len(x) > 3, words))

print(Greater_Than_Three_Letters)
```

```python
## Use zip to combine:

subjects = ["Math", "Physics", "Chemistry"]
marks = [90, 80, 95]

Combined_Subject_marks = list(zip(subjects,marks))

print(Combined_Subject_marks)
```

```python
## Use enumerate to print index + item for:

topics = ["python", "pandas", "numpy"]

for item, value in enumerate(topics):
    print(item, value)
```

```python
## Reading a text file 
try:
    with open ('data.txt','r') as f:
        content = f.read()
        print(content) 
except FileNotFoundError:
    print('File not found')
```

```python
## writing to a file 
try:
    with open('output.txt','w') as f:
        f.write('Hello Abhay')
except :
    print('Not File Created')
```

```python
## write a python list to a json file 
data = [1,2,3]
import json
try:
    with open('numbers.json','w') as f:
        json.dump(data,f,indent=4)
except FileNotFoundError:
    print('File not Created')
```
