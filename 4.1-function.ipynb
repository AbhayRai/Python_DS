Introduction to Functions

Definition:

A function is a block of code that performs a specific task. Functions help in organinzing code, reusing code, and improving readability.



```python
## syntax
def function_name(parameter):
    """Docstring"""
    # Function body
    return expression
```

```python
## why functions?
num=24
if num%2==0:
    print('The number is even')
else:
    print('The number is odd')
```

```python
def even_or_odd(num):
    """This function finds even or odd"""
    if(num%2==0):
        print('The number id even')
    else:
        print('The number is odd')
        
```

```python
## call this function 
even_or_odd(24)
```

```python
 ## function with multiple parameters
 
def multi_para(a,b):
    return a*b;
multi_para(5,6)
```

```python
## default Paramters

def def_para(name):
    print(f"Hello {name}")

def_para('Abhay')
```

```python
def def_para(name='Guest'):
    print(f"Hello {name}")

def_para()
```

```python
### Variable Length Arguements
## Positional and Keyword arguments

def print_numbers(*args):   ## *args: postional argument ## syntax
    print(args) ## convert into tuple 
    for number in args: ## iterate the tuple 
        print(number)
```

```python
print_numbers(1,2,3,4,4,5,'Abhay')
```

```python
### Keywords Arguments

def print_details(**kargs):     ## parameter are in key, value pair
    for key,value in kargs.items():
        print(f"{key}:{value}");

```

```python
# print_details(name='Abhay',age=25,Address='Assam')
print_details(name='Abhay',age=25,Address='Assam')
```

```python
def print_details(*args,**kargs):   ## positional argument follow after key arguments
    for number in args: ## iterate the tuple 
        print(number)
    for key,value in kargs.items():
        print(f"{key}:{value}");
```

```python
print_details(1,2,3,4,4,5,'Abhay',name='Abhay',age=25,Address='Assam')
```

```python
## Return statements

def multiply(a,b):
    return a*b;
multiply(2,3)
```
