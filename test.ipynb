# Sets in Python — Detailed Explanation

A **set** is an unordered collection of unique elements. Sets are mutable but cannot contain duplicate values.

## 1. Creating Sets



```python
# Create a set with curly braces
my_set = {1, 2, 3, 4, 5}
print(my_set)  # {1, 2, 3, 4, 5}

# Duplicates are automatically removed
my_set = {1, 2, 2, 3, 3, 4}
print(my_set)  # {1, 2, 3, 4}

# Create from a list using set()
my_set = set([1, 2, 3, 4, 5])
print(my_set)  # {1, 2, 3, 4, 5}

# Empty set (use set(), not {})
empty_set = set()
print(type(empty_set))  # <class 'set'>

# {} creates a dictionary, not a set
empty_dict = {}
print(type(empty_dict))  # <class 'dict'>
```



## 2. Adding & Removing Elements



```python
my_set = {1, 2, 3}

# Add single element
my_set.add(4)
print(my_set)  # {1, 2, 3, 4}

# Add duplicate (no effect)
my_set.add(3)
print(my_set)  # {1, 2, 3, 4}

# Remove by value (raises error if not found)
my_set.remove(3)
print(my_set)  # {1, 2, 4}

# Discard removes without error
my_set.discard(5)  # No error even if 5 doesn't exist
print(my_set)  # {1, 2, 4}

# Pop removes arbitrary element
popped = my_set.pop()
print(popped, my_set)

# Clear all elements
my_set.clear()
print(my_set)  # set()
```



## 3. Set Operations (Mathematical)



```python
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}

# Union — all elements from both sets
union = set1.union(set2)
print(union)  # {1, 2, 3, 4, 5, 6, 7, 8}

# Intersection — common elements
intersection = set1.intersection(set2)
print(intersection)  # {4, 5}

# Difference — elements in set1 but not in set2
difference = set1.difference(set2)
print(difference)  # {1, 2, 3}

# Symmetric difference — elements in either but not both
sym_diff = set1.symmetric_difference(set2)
print(sym_diff)  # {1, 2, 3, 6, 7, 8}
```



## 4. Set Comparison Methods



```python
set1 = {1, 2, 3, 4, 5}
set2 = {2, 3}
set3 = {1, 2, 3, 4, 5}

# Subset — all elements of set2 in set1?
print(set2.issubset(set1))  # True

# Superset — set1 contains all elements of set2?
print(set1.issuperset(set2))  # True

# Equality — same elements?
print(set1 == set3)  # True

# Disjoint — no common elements?
set4 = {6, 7, 8}
print(set1.isdisjoint(set4))  # True
```



## 5. Membership Test



```python
my_set = {1, 2, 3, 4, 5}

# Check if element exists
print(3 in my_set)  # True
print(10 in my_set)  # False
print(10 not in my_set)  # True
```



## 6. Iterating Over Sets



```python
my_set = {10, 20, 30, 40}

# Iterate with for loop
for item in my_set:
    print(item)

# Iterate with enumerate (get index too)
for index, item in enumerate(my_set):
    print(f"{index}: {item}")

# Convert to list to maintain order
my_list = sorted(my_set)
print(my_list)  # [10, 20, 30, 40]
```



## 7. Set Comprehension



```python
# Create a set using comprehension
squares = {x**2 for x in range(1, 6)}
print(squares)  # {1, 4, 9, 16, 25}

# Filter with condition
even_squares = {x**2 for x in range(1, 11) if x % 2 == 0}
print(even_squares)  # {4, 16, 36, 64, 100}
```



## 8. Key Differences: Set vs List vs Dict


```python
# LIST — ordered, allows duplicates
my_list = [1, 2, 2, 3]  # Duplicates kept
print(my_list[0])  # Can access by index

# SET — unordered, no duplicates
my_set = {1, 2, 2, 3}  # {1, 2, 3}
# print(my_set[0])  # Error! No indexing

# DICTIONARY — key-value pairs
my_dict = {"a": 1, "b": 2}
print(my_dict["a"])  # Access by key
```



## Summary of Set Methods

| Method | Purpose |
|--------|---------|
| `add()` | Add single element |
| `remove()` | Remove element (error if not found) |
| `discard()` | Remove element (no error if not found) |
| `pop()` | Remove arbitrary element |
| `clear()` | Remove all elements |
| `union()` | Combine sets |
| `intersection()` | Common elements |
| `difference()` | Elements in first set only |
| `symmetric_difference()` | Elements in either but not both |
| `issubset()` | Check if subset |
| `issuperset()` | Check if superset |
