## What is Series
#### A series is a one-dimensional labelled array. [Think of it as a single cloumn in excel]

```python
import pandas as pd
```

```python
## Creating a Series 

## From List
s = pd.Series(["a","b","c","d"],index=[1,2,3,4])
print(s)
```

```python
#### Access Value

display(s[1])
display(s[4])
```

```python
## From dictionary

d = pd.Series({1:'a',2:'b',3:'c'})
print(d)
```

```python
## Series Properties

print(type((s.values)))     ## <class 'numpy.ndarray'> [returns numpy array ]
type(list(s.values))        ## converted to list


print(type((s.index)))     ## <class 'pandas.core.indexes.base.Index'>
type(list(s.index))        ## converted to list

s.dtype        ## Doubt 

s.shape
```

## What is DataFrame?

#### A DataFrame is a two-dimensional table [Think of it like an excel sheet or SQL table]

```python
data = {
    'name': ['A','B','C','D'],      ## All arrays must be of the same length, or throws value error
    'age': [2,4,5,6]
}

df = pd.DataFrame(data)

print(df)
```

```python
## Creating DataFrame 

## From Dictionary

df_data = {
    'name':['A','B'],
    'age': [1,2]
}

df = pd.DataFrame(df_data)

df
```

```python
## From List of Lists

df_ListOfLists_data = [
    ['Name','Age'],
    ['A',1]
]

df = pd.DataFrame(df_ListOfLists_data)

df
```

```python

df_ListOfLists_data = [
    ['B',2],
    ['A',1]
]

df = pd.DataFrame(df_ListOfLists_data,columns=['Name','Age'])       ## Custom Column

df
```

```python
## List of Dictionaries

df_listofDictionaries_data =[
    {
        'Name': 'A',
        'Age' : 1
    },
    {
        'Name': 'B',
        'Age' : 2
    },
]

df = pd.DataFrame(df_listofDictionaries_data)

df
```

```python
## Accessing DataFrame Columns

df['Age']
```

```python
## Multiple Column

df[['Name','Age']]
```

```python
## DataFrame Properties

# df.shape

# df.columns

# type(df.columns)

df.index

df.dtypes
```

## Lesson 3 — Boolean Indexing & Filtering

### 1️⃣ What is Boolean Indexing?

### Boolean indexing means filtering rows using conditions.

```python
df = pd.DataFrame({
"name":["A","B","C","D"],
"age":[25,30,22,28],
"salary":[5000,6000,4500,7000]
})

df
```

```python
df["salary"] > 5000
```

```python
## This is a Boolean Series.

## Now apply it to the DataFrame.

df[df["salary"] > 5000]
```

```python
## Filtering with Multiple Conditions

## AND condition

df[(df['salary']>5000) & (df['age']>25)]
```

```python
## OR condition

df[(df['salary']>6500) | (df['age']<25)]
```

```python
## NOT Condition

df[~(df['salary']<=5000)]
```

```python
## df.loc[ row_condition , "name" ]

## So Pandas selects rows where the value is True.

## df["salary"] > 5000

## 0    False
## 1    True
## 2    False
## 3    True

## This is called a Boolean Series.

## Row 1 → True
## Row 3 → True
```
