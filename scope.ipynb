## Scope

```python
a = 10 # Global

def abc():
    a = 4 # local
    print(a)
abc()
```

```python
## To use the Global Variable inside a fn  we use the "global" keyword
a=10

def abc():
    global a
    a+=1
    print(f"The global Variable in fn: {a}")
print(f"The global Variable outside fn: {a}")    ## Before 
abc()       ## After
```

```python
## To use the Global Variable we use the "global" keyword
a=10

def abc():
    a=5
    a+=1
    print(f"The local Variable in fn: {a}")
print(f"The global Variable outside fn: {a}")    ## Before 
abc()
```

```python
## It is only for primitive data Types. Since for non-primitive we donot change the address

a = [1,2,3]

def non_primitive():
    a.append(4)
    print(f"The Non-Prmitive Value in the fn: {a}")
non_primitive()
```

```python
## Notes
## Python allows reading global variables inside functions, but requires the global keyword to modify them to avoid accidental side effects and scope confusion.
```
