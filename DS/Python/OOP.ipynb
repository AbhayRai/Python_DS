## OOP's

#### Inheritance

```python
## Type of Inheritance

## Single 
 
class A:
    def __init__(self,name,Address):
        self.name = name;
        self.Address = Address;
    def start(self):
        print(f"Class A Name: {self.name} Address: {self.Address}")
    
class B(A):
    def __init__(self,name,Address,Instance):
        super().__init__(name,Address)
        self.Instance = Instance;
    def start(self):
        super().start();
        print(f"Class B Instance: {self.Instance}")

B_Obj = B('Abhay','Bangalore','B')
B_Obj.start()
```

```python
## Multilevel Inheritance

class A:
    def __init__(self,name):
        self.name = name;
    def start(self):
        print(f"Class A Name: {self.name}");

class B(A):
    def __init__(self,name,Address):
        super().__init__(name);
        self.Address = Address;
    def start(self):
        super().start()
        print(f"Class B Address: {self.Address}")

class C(B):
    def __init__(self,name,Address,Instance):
        super().__init__(name,Address);
        self.Instance = Instance;
    def start(self):
        super().start()
        print(f"Class C Instance: {self.Instance}")

## Check the Inheritance Hierarchy 
print(C.mro())      ## [<class '__main__.C'>, <class '__main__.B'>, <class '__main__.A'>, <class 'object'>]

C_Obj = C('Abhay','Bangalore','C')
C_Obj.start()
```

```python
## Multiple Inheritance

class A:
    def __init__(self,name):
        self.name = name
    def start(self):
        print(f"Class A Name: {self.name}")

class B(A):
    def __init__(self,name,Address):
        super().__init__(name);
        self.Address = Address
    def start(self):
        super().start()
        print(f"Class B Address: {self.Address}")
    
class C(B,A):
    def __init__(self,name,Address):
        super().__init__(name,Address)
    def start(self):
        super().start()
        print(f"Class C")
    
## Check the Inheritance Hierarchy 
C.mro()     ## [__main__.C, __main__.B, __main__.A, object]

C_Obj = C('Abhay','Bangalore')
C_Obj.start()
```
